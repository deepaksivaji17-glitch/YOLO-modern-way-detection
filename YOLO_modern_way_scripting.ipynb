```python
import cv2
from ultralytics import YOLO

model = YOLO("yolov8n.pt")

video = cv2.VideoCapture(0)

while True:
    success, frame = video.read()
    if success == True:
        results = model(frame)
        print(results)
        
        for result in results:
            boxes = result.boxes
            for box in boxes:
                x1, y1, x2, y2 = map(int, box.xyxy[0])
                confidence = float(box.conf[0])
                class_id = int(box.cls[0])
                class_name = model.names[class_id]
                cv2.rectangle(frame,(x1, y1),(x2, y2),(0, 255, 0),3)
                label = f"{class_name} {confidence:.2f}"
                cv2.putText(frame,label,(x1, y1 - 10),cv2.FONT_HERSHEY_SIMPLEX,0.7,(0, 255, 0),2) # Put Label Above Box

        cv2.imshow("YOLO Webcam Detection", frame)
        key = cv2.waitKey(delay = 1)
        if key == 113 or key == 81:
            break
   
    else:
        print("Video Stopped")

video.release()
cv2.destroyAllWindows()
```

```python
import cv2
from ultralytics import YOLO

model=YOLO("yolov5nu.pt")

video=cv2.VideoCapture(0)

while True:
    success,frame=video.read()
    if success == True:
        results=model(frame)
        print(results)

        for result in results:
            boxes=result.boxes
            for box in boxes:
                x1,y1,x2,y2=map(int,box.xyxy[0])
                confidence=float(box.conf[0])
                class_id=int(box.cls[0])
                class_name=model.names[class_id]
                cv2.rectangle(img=frame,pt1=(x1,y1),pt2=(x2,y2),color=(0,255,0),thickness=3)
                label=f"{class_name} {confidence:.2f}"
                cv2.putText(img=frame, text=label, org=(x1, y1 - 10), fontFace=cv2.FONT_HERSHEY_SIMPLEX, fontScale=0.7, color=(0,255,0), thickness=2)


        cv2.imshow("YOLO Webcam Detection", frame)
        key = cv2.waitKey(delay=1)
        if key == 113 or key == 81:
            break
        else:
            print("Video stopped")


video.release()
cv2.destroyAllWindows()
                
```
